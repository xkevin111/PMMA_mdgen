```python
import torch
import torch.nn as nn

# 1. Define the parameters
# vocab_size: Number of unique categories (e.g., 20 amino acids + 1 mask = 21)
# embedding_dim: The size of the vector we want to represent each category
vocab_size = 21
embedding_dim = 16

# 2. Create the Embedding layer
# This creates a weight matrix of shape (21, 16)
embedding_layer = nn.Embedding(num_embeddings=vocab_size, embedding_dim=embedding_dim)

# 3. Define some input data (discrete indices)
# Let's say we have a batch of 2 sequences, each with 5 amino acids
# Indices must be within the range [0, vocab_size - 1]
input_indices = torch.tensor([
    [0, 5, 12, 20, 3],  # Sequence 1
    [19, 2, 20, 10, 1]  # Sequence 2
], dtype=torch.long)

# 4. Pass the indices through the embedding layer
# The layer acts as a lookup table
embedded_output = embedding_layer(input_indices)

# 5. Check the results
print(f"Input Shape (Indices): {input_indices.shape}")
print(f"Output Shape (Vectors): {embedded_output.shape}")

# Example: The vector for index '5' in the first sequence
print("\nVector for index 5 in the first sequence:")
print(embedded_output[0, 1])

weights = embedding_layer.weight.detach()

print("Embedding Weight Matrix Shape:", weights.shape)
print("\nWeight Matrix (The Lookup Table):")
print(weights)
```

```python
import torch
import torch.nn as nn

# 1. Define the parameters
# vocab_size: Number of unique categories (e.g., 20 amino acids + 1 mask = 21)
# embedding_dim: The size of the vector we want to represent each category
vocab_size = 1
embedding_dim = 16

# 2. Create the Embedding layer
# This creates a weight matrix of shape (21, 16)
embedding_layer = nn.Embedding(num_embeddings=vocab_size, embedding_dim=embedding_dim)

# 3. Define some input data (discrete indices)
# Let's say we have a batch of 2 sequences, each with 5 amino acids
# Indices must be within the range [0, vocab_size - 1]
input_indices = torch.tensor([
    [0, 0, 0, 0, 0],  # Sequence 1
    [0, 0, 0, 0, 0]  # Sequence 2
], dtype=torch.long)

# 4. Pass the indices through the embedding layer
# The layer acts as a lookup table
embedded_output = embedding_layer(input_indices)

# 5. Check the results
print(f"Input Shape (Indices): {input_indices.shape}")
print(f"Output Shape (Vectors): {embedded_output.shape}")

# Example: The vector for index '0' in the first sequence
print("\nVector for index 0 in the first sequence:")
print(embedded_output[0, 0])

weights = embedding_layer.weight.detach()

print("Embedding Weight Matrix Shape:", weights.shape)
print("\nWeight Matrix (The Lookup Table):")
print(weights)
```
